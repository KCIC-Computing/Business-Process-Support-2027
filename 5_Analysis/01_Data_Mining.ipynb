# Exploratory Data Analysis (EDA) vs Data Mining

## 1. Introduction

In data analytics, organisations work with large amounts of data to understand business performance, identify problems, discover patterns, and make informed decisions.

Two important activities used for these purposes are **Exploratory Data Analysis (EDA)** and **Data Mining**.

Although both involve analysing data and discovering useful insights, they have different primary purposes.

- **Exploratory Data Analysis (EDA)** focuses on understanding a dataset, examining its quality, exploring distributions, identifying relationships, and discovering unusual observations before drawing conclusions or building models.
- **Data Mining** focuses on discovering meaningful patterns, relationships, trends, and predictive rules from data using statistical, computational, and machine learning techniques.

In simple terms:

> **EDA helps us understand what is in the data, while data mining helps us discover useful patterns and relationships that can support business decisions.**

These activities overlap. EDA can reveal patterns, and data mining results often need to be explored and validated using EDA techniques.

---

## 2. What Is Exploratory Data Analysis (EDA)?

Exploratory Data Analysis is the process of investigating and summarising a dataset using statistical summaries, data visualisations, and analytical techniques to understand its structure and characteristics.

EDA helps an analyst answer questions such as:

- How many rows and columns are present?
- What does each column represent?
- Are there missing values or duplicate records?
- What are the average, minimum, and maximum values?
- How are numerical variables distributed?
- Are there any unusual observations or potential outliers?
- Are two or more variables related?
- Are there data quality problems that need to be addressed?

### Example: E-commerce Sales Data

Suppose an online shopping company provides a dataset containing the following columns:

| Order ID | Product | Quantity | Unit Price | Region | Sales |
|---|---|---:|---:|---|---:|
| 1001 | Laptop | 2 | 50,000 | South | 100,000 |
| 1002 | Mouse | 5 | 500 | North | 2,500 |
| 1003 | Keyboard | 3 | 1,500 | East | 4,500 |
| 1004 | Laptop | 1 | 50,000 | South | 50,000 |
| 1005 | Monitor | 2 | 12,000 | West | 24,000 |

Using EDA, an analyst could investigate:

1. Which product has the highest total sales?
2. Which region generates the most revenue?
3. What is the average order value?
4. Are there missing or incorrect prices?
5. Is sales revenue concentrated in a small number of products?

EDA helps the analyst understand the dataset before making business conclusions or building more advanced models.

### Common EDA Techniques

| Technique | Purpose | Python Example |
|---|---|---|
| Dataset inspection | Understand rows, columns, and data types | `df.info()` |
| Descriptive statistics | Summarise numerical variables | `df.describe()` |
| Missing-value analysis | Identify incomplete records | `df.isnull().sum()` |
| Duplicate detection | Find repeated records | `df.duplicated().sum()` |
| Frequency analysis | Understand category counts | `df["Region"].value_counts()` |
| Distribution analysis | Examine spread and shape | Histogram, box plot |
| Correlation analysis | Explore linear relationships | `df.corr(numeric_only=True)` |
| Grouped analysis | Compare categories | `df.groupby("Region")["Sales"].sum()` |

## 3. What Is Data Mining?

Data Mining is the process of extracting useful, previously unknown or non-obvious patterns, relationships, rules, and predictive information from datasets.

It often uses statistical methods, machine learning algorithms, pattern recognition, and computational techniques to uncover information that may not be immediately visible through basic summaries.

Data mining helps answer questions such as:

- Which customers are likely to stop using a service?
- Which products are frequently purchased together?
- Can customers be grouped according to their behaviour?
- Which factors are associated with loan defaults?
- Can future sales or customer demand be predicted?
- Which transactions appear unusual or potentially fraudulent?

### Example: Customer Purchase Behaviour

Suppose an online retailer has several years of transaction records covering thousands of customers.

The business wants to understand customer behaviour and improve marketing decisions.

Using data mining, the analyst could:

1. **Customer segmentation:** Group customers according to purchasing behaviour using clustering.
2. **Association rule mining:** Identify products that are frequently purchased together.
3. **Classification:** Predict whether a customer is likely to respond to a marketing campaign.
4. **Regression:** Estimate future customer spending.
5. **Anomaly detection:** Identify transactions that differ significantly from normal patterns.

These techniques can uncover complex patterns across large datasets.

### Common Data Mining Techniques

| Technique | Purpose | Example Algorithm or Method |
|---|---|---|
| Classification | Predict a categorical outcome | Decision Tree, Random Forest |
| Clustering | Group similar observations | K-Means |
| Regression | Predict a numerical value | Linear Regression |
| Association rule mining | Discover items that frequently occur together | Apriori |
| Anomaly detection | Identify unusual observations | Isolation Forest |
| Sequential pattern mining | Discover recurring sequences of events | Sequential pattern mining algorithms |

**Important:** Data mining does not always involve prediction. Association rule mining, clustering, and descriptive pattern discovery can reveal useful patterns without predicting a future outcome.

## 4. EDA vs Data Mining: Detailed Comparison

| Comparison Factor | Exploratory Data Analysis (EDA) | Data Mining |
|---|---|---|
| **Primary purpose** | Understand the dataset and explore its characteristics | Discover useful patterns, relationships, rules, or predictive information |
| **Main question** | What does the data look like, and what can we learn from exploring it? | What meaningful patterns or relationships can be extracted from the data? |
| **Typical starting point** | A dataset that needs to be inspected and understood | A dataset and a business question, discovery objective, or pattern-searching task |
| **Focus** | Data structure, quality, distributions, relationships, and unusual values | Patterns, segments, associations, classifications, predictions, and anomalies |
| **Typical methods** | Summary statistics, frequency tables, visualisations, correlation analysis | Machine learning, clustering, association rules, classification, regression |
| **Level of complexity** | Often begins with descriptive statistics and visual exploration; may include advanced analysis | Often involves algorithmic methods, parameter choices, and model evaluation |
| **Role of visualisation** | Central for understanding distributions and relationships | Useful for interpreting and communicating discovered patterns |
| **Role of algorithms** | May use simple statistical calculations or more advanced methods | Often uses algorithms designed to discover patterns or build models |
| **Use of labelled data** | Not required | Depends on the task; classification requires labelled examples for supervised training, while clustering does not |
| **Typical output** | Data summaries, plots, quality findings, hypotheses, and observations | Segments, association rules, detected anomalies, patterns, or trained predictive models |
| **Outlier handling** | Investigates unusual values and assesses whether they are errors or meaningful observations | May identify unusual records or use patterns that are sensitive to outliers |
| **Business value** | Helps establish a reliable understanding of the data and guide subsequent analysis | Helps uncover actionable relationships, customer groups, and predictive opportunities |
| **Evaluation** | Checks data quality, statistical behaviour, plausibility, and consistency | Depends on the task: predictive performance, rule usefulness, cluster quality, or anomaly validity |
| **Typical tools** | pandas, NumPy, Matplotlib, Seaborn, Excel | scikit-learn, mlxtend, SQL, Spark, and other analytics tools |
| **When performed** | Commonly early in an analysis and repeatedly throughout it | Often after initial data understanding, but can also be iterative and exploratory |
| **Main limitation** | Exploratory patterns may be misleading or may not generalise | Discovered patterns may be spurious, biased, unstable, or not useful in practice |

### Key distinction

EDA is mainly concerned with **understanding and investigating data**, whereas data mining is mainly concerned with **systematically discovering useful patterns or building models from data**.

However, they are not mutually exclusive. An EDA process can uncover meaningful patterns, and data mining frequently requires additional EDA to understand and validate its results.

## 5. Practical Example: EDA vs Data Mining

Imagine a supermarket wants to increase revenue and improve product recommendations.

It has a transaction dataset containing customer IDs, product names, order dates, quantities, and sales values.

### Stage A: Exploratory Data Analysis

The analyst explores the transaction data.

Questions:
- Which products generate the highest revenue?
- Which months have the highest sales?
- Are some product categories underperforming?
- Are there missing customer IDs?
- How does the quantity purchased vary across products?

Possible findings:
- Beverages generate the highest revenue.
- Weekend sales are higher than weekday sales.
- Some records contain missing product categories.
- A small number of transactions account for a large share of revenue.

These are descriptive findings. They help the analyst understand the data and decide which questions deserve further investigation.

### Stage B: Data Mining

The analyst investigates deeper patterns.

**Task 1: Product association analysis**

The analyst uses association rule mining to investigate whether customers who purchase bread also frequently purchase butter.

A possible rule is:

`Bread → Butter`

If the rule has sufficient support and confidence, the supermarket might consider product placement or bundled promotions. The rule does not prove that buying bread causes customers to buy butter.

**Task 2: Customer segmentation**

The analyst uses K-Means clustering to group customers based on features such as purchase frequency, average spending, and total spending.

Possible segments:
- Frequent, high-spending customers
- Occasional, high-value customers
- Frequent, low-spending customers
- Inactive or low-engagement customers

The business could use these segments to design more relevant marketing campaigns.

**Task 3: Sales prediction**

The analyst builds a model to estimate future sales based on historical sales, seasonality, prices, promotions, and other relevant variables.

The resulting predictions can help the supermarket plan inventory and staffing.

### Comparison of the Results

| EDA Result | Data Mining Result |
|---|---|
| Beverages generate the highest revenue. | A model identifies customer groups with different purchasing patterns. |
| Weekend sales are higher than weekday sales. | A predictive model estimates future sales using historical patterns. |
| Some product categories have missing values. | Association rules identify products frequently purchased together. |
| A few transactions account for a large proportion of revenue. | An anomaly detection algorithm flags unusual transactions for investigation. |

The EDA findings help establish what is happening in the dataset. Data mining can then investigate more complex patterns and build analytical solutions.

## 7. Python Demonstration: Data Mining with Clustering

We can extend the same dataset to demonstrate a simple data mining technique: clustering.

Clustering groups similar observations based on selected numerical features without requiring predefined class labels.

For this demonstration, we will group products using quantity sold and unit price. In a real business problem, these features and the level of analysis should be chosen carefully.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

features = df[["Quantity", "Unit_Price"]]

# Scale the features because they use different units and ranges
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# Create two clusters
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(scaled_features)

df[["Product", "Quantity", "Unit_Price", "Cluster"]]
```

### Explanation

1. **Feature selection:** We choose `Quantity` and `Unit_Price` as the inputs for clustering.
2. **Feature scaling:** Standardisation places the features on comparable scales so that unit price does not dominate merely because its numerical values are much larger.
3. **Model creation:** K-Means is configured to identify two clusters.
4. **Model fitting:** The algorithm groups records according to their distances from cluster centres.
5. **Result interpretation:** The `Cluster` column indicates the group assigned to each record.

**Important:** Cluster labels such as `0` and `1` are arbitrary identifiers. They do not represent good or bad performance. The clusters must be examined and interpreted before they can support a business decision.

This is a basic data mining demonstration because an algorithm is being used to discover groups in the data.

The result is illustrative, not a reliable customer segmentation model: the dataset is very small, and the features describe individual records rather than comprehensive customer behaviour.

## 8. How EDA and Data Mining Work Together

EDA and data mining are best understood as complementary activities in a data analytics workflow.

1. **Understand the business problem:** Define the question the organisation wants to answer.
2. **Collect the data:** Obtain data from databases, files, applications, or other sources.
3. **Perform EDA:** Inspect data structure, quality, distributions, and relationships.
4. **Prepare the data:** Handle missing values, duplicates, inconsistent formats, and relevant feature transformations.
5. **Apply data mining:** Use appropriate methods to discover patterns, segments, associations, or predictive relationships.
6. **Evaluate the results:** Check whether the findings are reliable and useful for the intended purpose.
7. **Communicate the findings:** Use visualisations, reports, and dashboards to explain the results.
8. **Support decisions:** Recommend actions based on the evidence and monitor their outcomes.

This workflow is iterative rather than strictly linear. If a mining algorithm produces unexpected results, the analyst may return to EDA, review the data, change the features, or reconsider the analytical approach.

**Advantages**

## 4. Create a Retail Transaction Dataset

````python
rng = np.random.default_rng(42)

n = 600
sales = pd.DataFrame({
    "transaction_id": np.arange(10001, 10001+n),
    "customer_id": rng.integers(1, 151, n),
    "region": rng.choice(["North", "South", "East", "West"], n),
    "category": rng.choice(["Electronics", "Office", "Home", "Grocery"], n,
                           p=[0.25, 0.30, 0.20, 0.25]),
    "quantity": rng.integers(1, 8, n),
    "unit_price": rng.choice([50, 75, 120, 250, 500, 900, 1500], n),
    "discount": rng.choice([0, 0.05, 0.10, 0.15], n,
                           p=[0.45, 0.30, 0.20, 0.05]),
    "payment": rng.choice(["UPI", "Card", "Cash"], n)
})

sales["revenue"] = sales["quantity"] * sales["unit_price"] * (1 - sales["discount"])
sales.head()
````

## 6. Anomaly Detection

### What is an anomaly?

An anomaly is an observation that behaves differently from the normal pattern.

Examples:

- unusually large transaction
- unusually high quantity
- unusually low or high price
- abnormal machine reading
- suspicious login behaviour

An anomaly is **a signal for investigation**, not proof of an error or fraud.

### 6.1 IQR-Based Detection

For a numerical variable:

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1
- Lower bound = Q1 − 1.5 × IQR
- Upper bound = Q3 + 1.5 × IQR

Observations outside the bounds are candidates for investigation.

````python

# 1. Calculate quartiles
q1 = sales["revenue"].quantile(0.25)
q3 = sales["revenue"].quantile(0.75)

# 2. Calculate the Interquartile Range (IQR)
iqr = q3 - q1

# 3. Calculate the lower and upper bounds
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

# 4. Identify revenue anomalies
sales["revenue_anomaly"] = ~sales["revenue"].between(lower, upper)

# 5. Display the calculated values
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower)
print("Upper bound:", upper)

# 6. Filter anomalous transactions
mask = sales["revenue_anomaly"].fillna(False)
anomalies = sales.loc[mask]

# 7. Sort by revenue in descending order and select the top 10
top_anomalies = anomalies.sort_values(
    by="revenue",
    ascending=False
).head(10)

# 8. Display the results
display(top_anomalies)
````

# 6.1 IQR-Based Outlier Detection

## 1. What Is IQR?

**IQR (Interquartile Range)** is a statistical measure used to understand the spread of the middle 50% of numerical data. It is also commonly used to identify potential outliers.

An **outlier** is an observation that is unusually low or high compared with the rest of the dataset.

For example, consider the following student marks:

`[10, 12, 14, 15, 16, 18, 20, 22, 24, 90]`

Most students scored between 10 and 24, but one student scored 90. The value 90 is much higher than the other observations, so we may want to investigate it.

The IQR method helps us determine whether a value falls outside the expected range calculated from the data.

---

## 2. Understanding Q1 — First Quartile

**Q1 is the first quartile, also known as the 25th percentile.** It represents the point below which approximately 25% of the observations fall.

To understand Q1, first arrange the data in ascending order:

`[10, 12, 14, 15, 16, 18, 20, 22, 24, 90]`

Using the median-of-halves method:

- Lower half: `[10, 12, 14, 15, 16]`
- The middle value of the lower half is `14`.

Therefore:

**Q1 = 14**

This means that the first quartile marks the boundary of the lower quarter of the dataset.

*Note:* Different statistical tools may calculate quartiles using different interpolation methods. Therefore, the exact value of Q1 can sometimes differ between methods.

---

## 3. Understanding Q3 — Third Quartile

**Q3 is the third quartile, also known as the 75th percentile.** It represents the point below which approximately 75% of the observations fall.

Using the same dataset:

`[10, 12, 14, 15, 16, 18, 20, 22, 24, 90]`

Using the median-of-halves method:

- Upper half: `[18, 20, 22, 24, 90]`
- The middle value of the upper half is `22`.

Therefore:

**Q3 = 22**

Q1 identifies the boundary of the lower quarter, while Q3 identifies the boundary of the upper quarter.

---

## 4. Calculating the Interquartile Range (IQR)

The IQR measures the spread of the middle 50% of the observations.

### Formula

\[
IQR = Q3 - Q1
\]

Where:

- Q1 = First quartile
- Q3 = Third quartile

### Example

From our student marks dataset:

- Q1 = 14
- Q3 = 22

Substituting the values:

\[
IQR = 22 - 14
\]

\[
\boxed{IQR = 8}
\]

**Interpretation:** The middle 50% of the data spans 8 marks, from Q1 = 14 to Q3 = 22.

Unlike the overall range, the IQR is relatively resistant to extreme values because it focuses on the middle portion of the data.

---

## 5. Calculating the Lower Bound

The lower bound is the threshold used to identify potential low outliers.

### Formula

\[
\text{Lower Bound} = Q1 - 1.5 \times IQR
\]

### Example

We know that:

- Q1 = 14
- IQR = 8

Substituting the values:

\[
\text{Lower Bound} = 14 - 1.5 \times 8
\]

First, calculate:

\[
1.5 \times 8 = 12
\]

Then:

\[
\text{Lower Bound} = 14 - 12
\]

\[
\boxed{\text{Lower Bound} = 2}
\]

**Interpretation:** Any observation below 2 is flagged as a potential low outlier.

For example, if a dataset contained a value of 0, that value would fall below the lower bound of 2 and would be flagged for investigation.

---

## 6. Calculating the Upper Bound

The upper bound is the threshold used to identify potential high outliers.

### Formula

\[
\text{Upper Bound} = Q3 + 1.5 \times IQR
\]

### Example

We know that:

- Q3 = 22
- IQR = 8

Substituting the values:

\[
\text{Upper Bound} = 22 + 1.5 \times 8
\]

First, calculate:

\[
1.5 \times 8 = 12
\]

Then:

\[
\text{Upper Bound} = 22 + 12
\]

\[
\boxed{\text{Upper Bound} = 34}
\]

**Interpretation:** Any observation above 34 is flagged as a potential high outlier.

For example, the value 90 is greater than 34, so it is flagged as a potential outlier.

---

## 7. Identifying Potential Outliers

Once the lower and upper bounds have been calculated, compare every observation with these boundaries.

The decision rules are:

- If a value is less than the lower bound, it is a potential low outlier.
- If a value is greater than the upper bound, it is a potential high outlier.
- If a value falls between the bounds, it is not flagged by the IQR rule.

### Example

For the student marks dataset:

`[10, 12, 14, 15, 16, 18, 20, 22, 24, 90]`

We calculated:

- Lower bound = 2
- Upper bound = 34

| Observation | Comparison | Result |
|---|---|---|
| 10 | Between 2 and 34 | Not flagged |
| 16 | Between 2 and 34 | Not flagged |
| 24 | Between 2 and 34 | Not flagged |
| 90 | Greater than 34 | Potential outlier |

Therefore, **90 is identified as a potential outlier** because it lies outside the calculated boundaries.

---

## 8. Why Do We Multiply IQR by 1.5?

The multiplier 1.5 is a conventional statistical rule used to identify observations that are unusually far from the middle 50% of the data.

It is widely used in boxplots and exploratory data analysis.

The multiplier creates a buffer around the central portion of the data:

- Lower bound = Q1 − 1.5 × IQR
- Upper bound = Q3 + 1.5 × IQR

Values outside these boundaries are flagged for further investigation.

However, **an outlier is not automatically an error**.

For example:
- A student scoring 90 may genuinely have performed exceptionally well.
- A shop recording unusually high sales may have experienced a successful promotional day.
- A height recorded as 1,750 cm instead of 175 cm may indicate a data-entry error.

The analyst should investigate the context before deciding whether to correct, retain or remove an observation.

## 9. Applying IQR-Based Detection Using Python

We can calculate Q1, Q3, IQR, the lower bound and the upper bound using pandas.

```python
import pandas as pd

# Create a numerical dataset
marks = pd.Series([
    10, 12, 14, 15, 16,
    18, 20, 22, 24, 90
])

# Calculate the first and third quartiles
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = marks[
    (marks < lower_bound) |
    (marks > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:")
print(outliers)
```

### Understanding the Python Code

1. `marks.quantile(0.25)` calculates the 25th percentile, or Q1.
2. `marks.quantile(0.75)` calculates the 75th percentile, or Q3.
3. `Q3 - Q1` calculates the IQR.
4. `Q1 - 1.5 * IQR` calculates the lower bound.
5. `Q3 + 1.5 * IQR` calculates the upper bound.
6. `(marks < lower_bound) | (marks > upper_bound)` identifies values outside either boundary. The `|` operator means **OR** in pandas Boolean expressions.
7. `marks[...]` filters the Series to return only the observations that meet the condition.

**Important:** pandas uses its default percentile interpolation method, which can produce different quartile values from the median-of-halves method used in the manual example. This is expected.

## 10. When Is IQR Useful in Data Analysis?

IQR-based detection can be applied to many numerical variables.

| Application | Example of a potential outlier |
|---|---|
| Education | An unusually high or low examination mark |
| Retail | An unusually large sales transaction |
| Finance | A transaction that differs substantially from typical values |
| Healthcare | An unusually high recorded measurement that requires checking |
| Data cleaning | A height or weight entered incorrectly |

IQR is particularly useful during **Exploratory Data Analysis (EDA)**, when analysts investigate distributions, detect unusual observations and assess data quality before further analysis.

### Key Takeaways

- **Q1:** The 25th percentile.
- **Q3:** The 75th percentile.
- **IQR:** The difference between Q3 and Q1; it measures the spread of the middle 50% of the data.
- **Lower bound:** Q1 − 1.5 × IQR.
- **Upper bound:** Q3 + 1.5 × IQR.
- **Potential outlier:** An observation below the lower bound or above the upper bound.

**Remember:** The IQR method identifies unusual observations for investigation. It does not prove that those observations are incorrect, and they should not be removed automatically.

## 8. Customer Behaviour and Simple Segmentation

A common business mining task is identifying groups of customers by behaviour.

Useful measures:

- **Recency** — how recently a customer purchased
- **Frequency** — how often a customer purchased
- **Monetary value** — how much a customer spent

For Unit 17, focus on the idea of segmentation rather than implementing machine-learning clustering. Full clustering algorithms can be taught in the next unit.

````python
customer_summary = (
    sales.groupby("customer_id")
         .agg(
             purchase_count=("transaction_id", "count"),
             total_spend=("revenue", "sum"),
             average_order_value=("revenue", "mean")
         )
         .reset_index()
)

customer_summary["customer_value_band"] = pd.cut(
    customer_summary["total_spend"],
    bins=[
        -np.inf,
        customer_summary["total_spend"].quantile(0.33),
        customer_summary["total_spend"].quantile(0.66),
        np.inf
    ],
    labels=["Low", "Medium", "High"]
)

customer_summary.head(10)
````

````python
customer_summary["customer_value_band"].value_counts().sort_index()
````

## 9. Simple Market-Basket / Co-Occurrence Concept

Another classic data-mining problem is:

> Which products or categories tend to occur together?

### Key ideas

- **Support:** how frequently an item/set appears.
- **Confidence:** how often B appears when A appears.
- **Lift:** how much stronger the relationship is than a baseline expectation.

For this unit, students should understand the concepts and business meaning. A full association-rule implementation can be left to a more specialised analytics/ML unit.

````python
transactions = [
    {"Bread", "Butter", "Milk"},
    {"Bread", "Butter"},
    {"Bread", "Milk"},
    {"Bread", "Butter", "Jam"},
    {"Milk", "Cookies"},
    {"Bread", "Butter", "Milk"},
    {"Bread", "Jam"},
    {"Butter", "Milk"},
]

def support(itemset, transactions):
    itemset = set(itemset)
    return sum(itemset.issubset(t) for t in transactions) / len(transactions)

candidate_pairs = [
    ("Bread", "Butter"),
    ("Bread", "Milk"),
    ("Butter", "Milk"),
    ("Bread", "Jam"),
]

for pair in candidate_pairs:
    print(pair, "support =", round(support(pair, transactions), 3))
````
