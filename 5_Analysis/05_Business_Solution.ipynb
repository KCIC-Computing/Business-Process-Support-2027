## 2. Business Problem vs Technical Problem

### Business problem

> Customer repeat purchases are falling.

### Technical question

> Which customer behaviours are associated with low repeat-purchase activity?

### Data-science task

> Collect, prepare and analyse customer transaction data to identify patterns and produce recommendations.

Students should always start with the **business problem**, not the code.

## 3. Define the Problem Clearly

A good problem statement should identify:

- organisation/context
- problem
- affected area
- desired outcome
- time period or scope

### Weak

> Analyse customer data.

### Strong

> An e-commerce business has seen a decline in repeat purchases over the last six months and wants to identify customer groups with reduced activity and the factors associated with the decline.

## 4. Define SMART Objectives

Students can use:

- Specific
- Measurable
- Achievable
- Relevant
- Time-bound

Example:

> Analyse six months of customer transaction data to identify segments with low repeat-purchase activity and recommend actions that could support retention.

## 6. Requirements

### Business requirements

What the organisation needs to achieve.

### User requirements

What a user needs from the solution.

### Data requirements

What data is needed.

### System requirements

What the system/tool must provide.

Example:

| Type | Requirement |
|---|---|
| Business | Improve retention decision making |
| User | Manager can identify low-activity customers |
| Data | Customer ID, date, transaction value and category |
| System | Generate repeatable monthly analysis |

## 7. KPI Definition

Possible KPIs:

- revenue
- order count
- average order value
- repeat-purchase rate
- active customers

- forecast error

### Important

A KPI is useful only when:

1. it relates to the business objective

3. the result can influence a decision.

## 11. Data Quality Risks

A good design explicitly identifies risks.

Examples:

- missing values
- duplicate records
- incorrect dates
- inconsistent categories
- delayed data
- inaccurate source systems
- biased samples
- external data changes

### Risk principle

> **A sophisticated analysis built on poor data can still produce poor decisions.**

## 12. Example End-to-End Case

### Business

An online retailer wants to understand why sales have become more volatile and whether competitor prices may be affecting performance.

### Available data

- internal sales CSV
- product master data
- competitor price data from an approved source
- monthly sales history

### Goal

Produce evidence and recommendations for pricing and inventory decisions.

````python
# Create a small case-study dataset
case_sales = pd.DataFrame({
    "month": pd.date_range("2025-01-01", periods=12, freq="MS"),
    "sales": [120, 125, 130, 128, 135, 142, 139, 136, 145, 151, 160, 168],
    "competitor_avg_price_index": [100, 99, 98, 99, 97, 96, 97, 98, 96, 95, 94, 94]
})

case_sales
````

## 13. Solution Workflow for the Case

Students can combine:

1. data validation
2. EDA
3. relationship analysis
4. simple forecasting
5. external price comparison
6. visualisation
7. recommendation

Notice that no machine-learning model is required to demonstrate the idea of an end-to-end data-science solution.

````python
case_sales["sales_change_pct"] = case_sales["sales"].pct_change() * 100
case_sales["price_change_pct"] = case_sales["competitor_avg_price_index"].pct_change() * 100

case_sales
````

## 15. Testing

Do not finish when the code runs.

Test:

- inputs
- missing data
- incorrect types
- unexpected values
- outputs
- calculations
- visualisations
- requirement fulfilment

A simple test table:

| Requirement | Test | Expected | Result |
|---|---|---|---|
| Load sales data | Valid CSV | DataFrame created | Pass |
| Handle missing value | Insert missing price | Missing value flagged | Pass |
| Calculate revenue | Known example | Correct value | Pass |
| Produce output | Run report | Report generated | Pass |

````python
test_cases = pd.DataFrame({
    "requirement": [
        "Load sales data",
        "Handle missing price",
        "Calculate revenue",
        "Generate output"
    ],
    "test": [
        "Provide valid CSV",
        "Insert missing price",
        "Use known quantity × price example",
        "Run analysis end-to-end"
    ],
    "expected": [
        "DataFrame created",
        "Value detected/flagged",
        "Correct calculation",
        "Output produced"
    ],
    "status": ["Pass", "Pass", "Pass", "Pass"]
})

test_cases
````

## 16. Evaluate Against Requirements

For D2-style evaluation, students should compare the solution against:

### User requirements
Can the user understand and use it?

### Business requirements
Does it support the original decision?

### Data quality
Is the evidence reliable enough?

### Technical feasibility
Can it be repeated and maintained?

### Risks
What could make the result wrong or misleading?

## 17. Recommendation Framework

A justified recommendation should contain:

**Evidence → Meaning → Business impact → Action → Expected benefit → Risk/limitation**

Example:

> Competitor prices were consistently below our prices for two key products. This indicates that our current pricing may reduce competitiveness in those products. Management should review pricing for these products while checking margin constraints. The recommendation should be monitored using sales volume and margin KPIs.

## 20. Final LO4 Checklist
